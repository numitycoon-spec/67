```python
import numpy as np
import pandas as pd

data = pd.read_csv("train_dataset.csv")
test = pd.read_csv("/content/test_without_gt (2).csv")
template = pd.read_csv("/content/submission_template_for_test (1) (1).csv")

print(data.shape, test.shape, template.shape)

y = data["12-Month Default Flag"]
X = data.drop("12-Month Default Flag", axis=1)
```

```python
def add_features(df):
    df = df.copy()
    df["age_years"] = df["Age (in months)"] / 12
    df["residence_years"] = df["Residence Period (in months)"] / 12
    df["residence_to_age"] = df["Residence Period (in months)"] / df["Age (in months)"]
    df["limit_to_age"] = df["Desired Credit Limit"] / df["Age (in months)"]
    df["rent_to_limit"] = df["Rent Burden"] / (df["Desired Credit Limit"] + 1)

    inc = df["Declared Annual Income"] + 1
    df["limit_to_income"] = df["Desired Credit Limit"] / inc
    df["borrow_to_income"] = df["Declared Borrowing Amount"] / inc
    df["rent_to_income"] = df["Rent Burden"] / inc
    df["avg_per_borrowing"] = df["Declared Borrowing Amount"] / (df["Declared Number of Borrowings"] + 1)
    df["total_debt_after"] = (df["Declared Borrowing Amount"] + df["Desired Credit Limit"]) / inc
    df["income_per_age"] = df["Declared Annual Income"] / df["Age (in months)"]
    df["borrow_to_limit"] = df["Declared Borrowing Amount"] / (df["Desired Credit Limit"] + 1)
    return df.replace([np.inf, -np.inf], np.nan)

X2 = add_features(X).drop(columns=["ID"])
test2 = add_features(test).drop(columns=["ID"])

cat_cols = [c for c in ["Application Channel", "Gender", "Marital Status",
                        "Occupation Category", "Industry Category",
                        "Employment Type Category", "Insurance Type",
                        "Residence Type", "Residence Ownership Status"]
            if c in X2.columns]
print("categorical columns:", cat_cols)
```

```python
def add_stat_features(train, test, cat_cols, num_cols):
    train, test = train.copy(), test.copy()
    both = pd.concat([train, test], axis=0)

    for c in cat_cols:
        freq = both[c].astype(str).value_counts(normalize=True)
        train[c + "_freq"] = train[c].astype(str).map(freq).astype(float)
        test[c + "_freq"] = test[c].astype(str).map(freq).astype(float)

    for g in ["Occupation Category", "Industry Category"]:
        if g not in both.columns:
            continue
        for n in num_cols:
            mean = both.groupby(g)[n].mean()
            train[f"{n}_mean_by_{g}"] = train[g].map(mean).astype(float)
            test[f"{n}_mean_by_{g}"] = test[g].map(mean).astype(float)
            train[f"{n}_diff_{g}"] = train[n] - train[f"{n}_mean_by_{g}"]
            test[f"{n}_diff_{g}"] = test[n] - test[f"{n}_mean_by_{g}"]
    return train, test

num_cols = [c for c in ["Desired Credit Limit", "Age (in months)", "Rent Burden",
                        "Declared Annual Income", "Declared Borrowing Amount",
                        "limit_to_income", "borrow_to_income"]
            if c in X2.columns]
X3, test3 = add_stat_features(X2, test2, cat_cols, num_cols)
print(X3.shape, test3.shape)
```

```python
X_lgb, test_lgb = X3.copy(), test3.copy()
for c in cat_cols:
    cats = pd.concat([X3[c], test3[c]]).astype("category").cat.categories
    X_lgb[c] = pd.Categorical(X3[c], categories=cats)
    test_lgb[c] = pd.Categorical(test3[c], categories=cats)

X_cb, test_cb = X3.copy(), test3.copy()
X_cb[cat_cols] = X_cb[cat_cols].astype(str)
test_cb[cat_cols] = test_cb[cat_cols].astype(str)
```

```python
import lightgbm as lgb
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score
from scipy.stats import rankdata

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=1)

oof_lgb = np.zeros(len(X3)); test_lgb_pred = np.zeros(len(test3))
oof_cb = np.zeros(len(X3)); test_cb_pred = np.zeros(len(test3))

for fold, (tr, va) in enumerate(skf.split(X3, y)):
    m1 = LGBMClassifier(n_estimators=3000, learning_rate=0.01, num_leaves=15,
                        min_child_samples=100, subsample=0.8, subsample_freq=1,
                        colsample_bytree=0.7, reg_lambda=5,
                        random_state=1, verbose=-1)
    m1.fit(X_lgb.iloc[tr], y.iloc[tr],
           eval_set=[(X_lgb.iloc[va], y.iloc[va])], eval_metric="auc",
           callbacks=[lgb.early_stopping(100, verbose=False)])
    oof_lgb[va] = m1.predict_proba(X_lgb.iloc[va])[:, 1]
    test_lgb_pred += m1.predict_proba(test_lgb)[:, 1] / 5

    m2 = CatBoostClassifier(iterations=3000, learning_rate=0.03, depth=6,
                            eval_metric="AUC", early_stopping_rounds=100,
                            cat_features=cat_cols, random_seed=1, verbose=0)
    m2.fit(X_cb.iloc[tr], y.iloc[tr], eval_set=(X_cb.iloc[va], y.iloc[va]))
    oof_cb[va] = m2.predict_proba(X_cb.iloc[va])[:, 1]
    test_cb_pred += m2.predict_proba(test_cb)[:, 1] / 5

    print(f"fold {fold}: lgb={roc_auc_score(y.iloc[va], oof_lgb[va]):.5f}  "
          f"cat={roc_auc_score(y.iloc[va], oof_cb[va]):.5f}")

print("LGB  CV AUC:", roc_auc_score(y, oof_lgb))
print("Cat  CV AUC:", roc_auc_score(y, oof_cb))
```

```python
def rank(a):
    return rankdata(a) / len(a)

blend_oof = (rank(oof_lgb) + rank(oof_cb)) / 2
print("Blend CV AUC:", roc_auc_score(y, blend_oof))

final = (rank(test_lgb_pred) + rank(test_cb_pred)) / 2
```

```python
from xgboost import XGBClassifier

oof_list = [oof_lgb, oof_cb]
test_list = [test_lgb_pred, test_cb_pred]

# LightGBM with 3 more seeds
for seed in [2, 3, 4]:
    skf_s = StratifiedKFold(n_splits=5, shuffle=True, random_state=seed)
    o = np.zeros(len(X3)); p = np.zeros(len(test3))
    for tr, va in skf_s.split(X3, y):
        m = LGBMClassifier(n_estimators=3000, learning_rate=0.01, num_leaves=15,
                           min_child_samples=100, subsample=0.8, subsample_freq=1,
                           colsample_bytree=0.7, reg_lambda=5,
                           random_state=seed, verbose=-1)
        m.fit(X_lgb.iloc[tr], y.iloc[tr],
              eval_set=[(X_lgb.iloc[va], y.iloc[va])], eval_metric="auc",
              callbacks=[lgb.early_stopping(100, verbose=False)])
        o[va] = m.predict_proba(X_lgb.iloc[va])[:, 1]
        p += m.predict_proba(test_lgb)[:, 1] / 5
    print("lgb seed", seed, roc_auc_score(y, o))
    oof_list.append(o); test_list.append(p)

# XGBoost
o = np.zeros(len(X3)); p = np.zeros(len(test3))
for tr, va in skf.split(X3, y):
    m = XGBClassifier(n_estimators=3000, learning_rate=0.01, max_depth=4,
                      min_child_weight=20, subsample=0.8, colsample_bytree=0.7,
                      reg_lambda=5, tree_method="hist", enable_categorical=True,
                      eval_metric="auc", early_stopping_rounds=100, random_state=1)
    m.fit(X_lgb.iloc[tr], y.iloc[tr],
          eval_set=[(X_lgb.iloc[va], y.iloc[va])], verbose=False)
    o[va] = m.predict_proba(X_lgb.iloc[va])[:, 1]
    p += m.predict_proba(test_lgb)[:, 1] / 5
print("xgb", roc_auc_score(y, o))
oof_list.append(o); test_list.append(p)
```

```python
blend_oof = sum(rank(o) for o in oof_list) / len(oof_list)
print("Blend of", len(oof_list), "models CV AUC:", roc_auc_score(y, blend_oof))

final = sum(rank(p) for p in test_list) / len(test_list)
template["12-Month Default Flag"] = final
template.to_csv("submission4.csv", index=False)

from google.colab import files
files.download("submission4.csv")
```

```python
def cv_lgb(Xd, seed=1):
    skf_t = StratifiedKFold(n_splits=5, shuffle=True, random_state=seed)
    o = np.zeros(len(Xd))
    for tr, va in skf_t.split(Xd, y):
        m = LGBMClassifier(n_estimators=3000, learning_rate=0.02, num_leaves=15,
                           min_child_samples=100, subsample=0.8, subsample_freq=1,
                           colsample_bytree=0.7, reg_lambda=5,
                           random_state=seed, verbose=-1)
        m.fit(Xd.iloc[tr], y.iloc[tr],
              eval_set=[(Xd.iloc[va], y.iloc[va])], eval_metric="auc",
              callbacks=[lgb.early_stopping(100, verbose=False)])
        o[va] = m.predict_proba(Xd.iloc[va])[:, 1]
    return roc_auc_score(y, o)

print("baseline:", cv_lgb(X_lgb))

extra = [c for c in ["Company Size Category", "Payday"] if c in X3.columns]
X_try = X_lgb.copy()
for c in extra:
    cats = pd.concat([X3[c], test3[c]]).astype("category").cat.categories
    X_try[c] = pd.Categorical(X3[c], categories=cats)
print("with", extra, "as categorical:", cv_lgb(X_try))
```

```python
def add_stat_features2(train, test, cat_cols, num_cols, groups):
    train, test = train.copy(), test.copy()
    both = pd.concat([train, test], axis=0)

    for c in cat_cols:
        freq = both[c].astype(str).value_counts(normalize=True)
        train[c + "_freq"] = train[c].astype(str).map(freq).astype(float)
        test[c + "_freq"] = test[c].astype(str).map(freq).astype(float)

    for g in groups:
        for n in num_cols:
            mean = both.groupby(g)[n].mean()
            train[f"{n}_mean_by_{g}"] = train[g].map(mean).astype(float)
            test[f"{n}_mean_by_{g}"] = test[g].map(mean).astype(float)
            train[f"{n}_diff_{g}"] = train[n] - train[f"{n}_mean_by_{g}"]
            test[f"{n}_diff_{g}"] = test[n] - test[f"{n}_mean_by_{g}"]
    return train, test

groups2 = [g for g in ["Occupation Category", "Industry Category",
                       "Company Size Category", "Employment Type Category"]
           if g in X2.columns]
num_cols2 = [c for c in ["Desired Credit Limit", "Age (in months)", "Rent Burden",
                         "Declared Annual Income", "Declared Borrowing Amount",
                         "Declared Number of Borrowings", "Residence Period (in months)",
                         "limit_to_income", "borrow_to_income", "rent_to_income",
                         "total_debt_after"]
             if c in X2.columns]

X4, test4 = add_stat_features2(X2, test2, cat_cols, num_cols2, groups2)

X_try2 = X4.copy()
for c in cat_cols:
    cats = pd.concat([X4[c], test4[c]]).astype("category").cat.categories
    X_try2[c] = pd.Categorical(X4[c], categories=cats)

print("features:", X_try2.shape[1])
print("baseline:", cv_lgb(X_lgb))
print("more group stats:", cv_lgb(X_try2))
```

```python
# data for the new feature set
test_lgb4 = test4.copy()
for c in cat_cols:
    cats = pd.concat([X4[c], test4[c]]).astype("category").cat.categories
    test_lgb4[c] = pd.Categorical(test4[c], categories=cats)

X_cb4, test_cb4 = X4.copy(), test4.copy()
X_cb4[cat_cols] = X_cb4[cat_cols].astype(str)
test_cb4[cat_cols] = test_cb4[cat_cols].astype(str)

oof_list2 = list(oof_list)
test_list2 = list(test_list)

# LightGBM, 3 seeds
for seed in [1, 2, 3]:
    skf_s = StratifiedKFold(n_splits=5, shuffle=True, random_state=seed)
    o = np.zeros(len(X4)); p = np.zeros(len(test4))
    for tr, va in skf_s.split(X4, y):
        m = LGBMClassifier(n_estimators=3000, learning_rate=0.01, num_leaves=15,
                           min_child_samples=100, subsample=0.8, subsample_freq=1,
                           colsample_bytree=0.7, reg_lambda=5,
                           random_state=seed, verbose=-1)
        m.fit(X_try2.iloc[tr], y.iloc[tr],
              eval_set=[(X_try2.iloc[va], y.iloc[va])], eval_metric="auc",
              callbacks=[lgb.early_stopping(100, verbose=False)])
        o[va] = m.predict_proba(X_try2.iloc[va])[:, 1]
        p += m.predict_proba(test_lgb4)[:, 1] / 5
    print("new lgb seed", seed, roc_auc_score(y, o))
    oof_list2.append(o); test_list2.append(p)

# CatBoost
o = np.zeros(len(X4)); p = np.zeros(len(test4))
for tr, va in skf.split(X4, y):
    m = CatBoostClassifier(iterations=3000, learning_rate=0.03, depth=6,
                           eval_metric="AUC", early_stopping_rounds=100,
                           cat_features=cat_cols, random_seed=1, verbose=0)
    m.fit(X_cb4.iloc[tr], y.iloc[tr], eval_set=(X_cb4.iloc[va], y.iloc[va]))
    o[va] = m.predict_proba(X_cb4.iloc[va])[:, 1]
    p += m.predict_proba(test_cb4)[:, 1] / 5
print("new cat", roc_auc_score(y, o))
oof_list2.append(o); test_list2.append(p)

# blend everything
blend_oof = sum(rank(o) for o in oof_list2) / len(oof_list2)
print("Blend of", len(oof_list2), "models CV AUC:", roc_auc_score(y, blend_oof))

final = sum(rank(p) for p in test_list2) / len(test_list2)
template["12-Month Default Flag"] = final
template.to_csv("submission5.csv", index=False)

from google.colab import files
files.download("submission5.csv")
```
